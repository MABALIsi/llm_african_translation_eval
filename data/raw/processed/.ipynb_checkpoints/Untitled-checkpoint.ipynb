```python
import pandas as  pd
data = pd.read_csv(
    "../data/processed/processed_lingala.csv"
)

data.drop_duplicates(inplace=True)

data.dropna(inplace=True)

data["french"] = data["french"].str.lower()

data["lingala"] = data["lingala"].str.lower()

data["length"] = data["french"].apply(len)

data.head()

data.to_csv(
    "../data/processed/clean_lingala.csv",
    index=False
)
```
