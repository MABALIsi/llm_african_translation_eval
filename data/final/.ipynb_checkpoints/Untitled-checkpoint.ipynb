```python
import json, os
from pathlib import Path

FINAL_DIR = Path("C:/Users/Albin Mabali/llm_african_translation/data/final")
FINAL_DIR.mkdir(parents=True, exist_ok=True)

# ── Phrases supplémentaires (650 nouvelles par langue) ──
phrases_supplementaires = {
    "sante": [
        "Le médecin ausculte le patient soigneusement.",
        "L'infirmière prend la température du malade.",
        "Le bébé est en bonne santé à la naissance.",
        "Les antibiotiques combattent les infections bactériennes.",
        "La clinique est équipée de matériel moderne.",
        "Le pharmacien explique comment prendre le médicament.",
        "Les douleurs thoraciques nécessitent une consultation urgente.",
        "La prévention du choléra passe par l'hygiène.",
        "Les enfants sont pesés chaque mois au dispensaire.",
        "La fièvre jaune se prévient par la vaccination.",
        "Le sang contaminé transmet de nombreuses maladies.",
        "L'opération chirurgicale a été un succès total.",
        "Les médicaments traditionnels complètent la médecine moderne.",
        "La tuberculose se soigne avec des antibiotiques spéciaux.",
        "Le dispensaire manque de personnel qualifié.",
        "Les soins prénataux réduisent la mortalité maternelle.",
        "L'hypertension touche de nombreux adultes.",
        "Le diabète nécessite un régime alimentaire strict.",
        "La rougeole est évitable grâce au vaccin.",
        "Les moustiques transmettent le paludisme la nuit.",
        "La déshydratation est dangereuse pour les nourrissons.",
        "Le centre de santé ouvre à huit heures.",
        "Les agents de santé visitent les villages régulièrement.",
        "La malaria cause beaucoup de décès chez les enfants.",
        "Les lunettes corrigent les problèmes de vision.",
        "L'accouchement à domicile présente des risques.",
        "Les dents doivent être brossées deux fois par jour.",
        "La pneumonie se traite avec des antibiotiques adaptés.",
        "Les soins palliatifs améliorent la qualité de vie.",
        "L'alcool en excès détruit le foie progressivement.",
    ],
    "education": [
        "Les élèves mémorisent les tables de multiplication.",
        "Le professeur corrige les copies des étudiants.",
        "La récréation dure vingt minutes le matin.",
        "Les enfants apprennent à compter jusqu'à cent.",
        "La rentrée scolaire est fixée au premier octobre.",
        "Les bourses permettent aux pauvres d'accéder à l'université.",
        "L'enseignement bilingue est pratiqué dans certaines écoles.",
        "Les parents s'impliquent dans l'éducation de leurs enfants.",
        "Le tableau blanc remplace progressivement le tableau noir.",
        "Les ordinateurs facilitent l'apprentissage des élèves.",
        "L'école maternelle prépare les enfants à la primaire.",
        "Les activités parascolaires développent les talents des élèves.",
        "La lecture quotidienne améliore le vocabulaire.",
        "Les examens nationaux évaluent le niveau des élèves.",
        "Le cartable de l'élève contient ses livres et cahiers.",
        "Les sciences enseignent à observer et comprendre le monde.",
        "La géographie décrit les pays et leurs ressources.",
        "Les langues nationales sont enseignées à l'école.",
        "Le baccalauréat ouvre la porte à l'université.",
        "Les enseignants se forment pendant les vacances.",
        "La scolarisation des filles est une priorité nationale.",
        "Les internats hébergent les élèves venus de loin.",
        "Le sport développe la santé et l'esprit d'équipe.",
        "Les cours particuliers aident les élèves en difficulté.",
        "L'enseignement technique forme des professionnels qualifiés.",
        "La chimie expérimentale se pratique en laboratoire.",
        "Les devoirs à la maison consolident les apprentissages.",
        "Le conseil des parents se réunit chaque trimestre.",
        "Les notes sont communiquées aux parents régulièrement.",
        "L'éducation religieuse est enseignée dans certaines écoles.",
    ],
    "quotidien": [
        "Je me réveille à six heures chaque matin.",
        "Le café chaud me donne de l'énergie.",
        "Je prends une douche avant d'aller travailler.",
        "Les embouteillages ralentissent la circulation en ville.",
        "Ma femme prépare un délicieux repas ce soir.",
        "Les enfants rentrent de l'école à quatorze heures.",
        "Nous regardons le journal télévisé chaque soir.",
        "Je lis un livre avant de dormir.",
        "Le marché du quartier est ouvert tôt le matin.",
        "Mon collègue m'a prêté de l'argent.",
        "La mosquée appelle les fidèles à la prière.",
        "Nous partageons le repas en famille.",
        "Je rends visite à ma mère chaque dimanche.",
        "Le vendeur m'a proposé un bon prix.",
        "La nuit est fraîche et agréable.",
        "Les voisins organisent une fête ce weekend.",
        "Je recharge mon téléphone le soir.",
        "Les enfants regardent des dessins animés.",
        "Le repas de midi est servi à treize heures.",
        "Je me couche tôt pour être en forme.",
        "La radio diffuse de la musique le matin.",
        "Le vélo est mon moyen de transport quotidien.",
        "Je bois deux litres d'eau par jour.",
        "Les courses du week-end prennent une heure.",
        "Mon fils aide à préparer le dîner.",
        "Nous jouons aux cartes après le repas.",
        "Le barbier coupe les cheveux dans le quartier.",
        "J'envoie de l'argent à ma famille au village.",
        "La nuit, les étoiles brillent dans le ciel.",
        "Le matin, les oiseaux chantent dans les arbres.",
    ],
    "agriculture": [
        "Le tracteur laboure vingt hectares par jour.",
        "Les semences hybrides produisent de meilleurs rendements.",
        "Le paysan construit un grenier pour stocker le mil.",
        "La pêche artisanale nourrit les familles côtières.",
        "Les abeilles pollinisent les fleurs des vergers.",
        "Le maraîchage emploie beaucoup de femmes rurales.",
        "La banane plantain est cultivée dans tout le pays.",
        "Les engrais chimiques améliorent la productivité.",
        "Le fleuve irrigue les champs pendant la sécheresse.",
        "Les outils agricoles facilitent le travail des paysans.",
        "La culture du riz nécessite beaucoup d'eau.",
        "Les poulets sont élevés dans des poulaillers modernes.",
        "Le manioc est la principale source de nourriture.",
        "La forêt fournit du bois pour cuisiner.",
        "Les coopératives vendent les récoltes à bon prix.",
        "Le moulin transforme le mil en farine.",
        "Les femmes trient et nettoient le riz récolté.",
        "Le bœuf de labour aide à préparer les champs.",
        "Les insectes nuisibles détruisent les récoltes.",
        "La jachère permet à la terre de se reposer.",
        "Les pluies abondantes favorisent de bonnes récoltes.",
        "Le paysan vend ses légumes au marché hebdomadaire.",
        "La calebasse est utilisée pour transporter l'eau.",
        "Les vaches laitières produisent du lait frais.",
        "Les terres dégradées nécessitent une reforestation urgente.",
        "Le jardin potager alimente la famille en légumes.",
        "La traction animale est répandue dans les villages.",
        "Les techniciens agricoles conseillent les paysans.",
        "La récolte de l'arachide se fait en saison sèche.",
        "Le paysan économise pour acheter de meilleurs outils.",
    ],
    "administration": [
        "La carte d'identité nationale est valable dix ans.",
        "Le bureau d'état civil enregistre les naissances.",
        "Le tribunal civil règle les conflits fonciers.",
        "La gendarmerie assure la sécurité en zone rurale.",
        "Le budget municipal finance les infrastructures locales.",
        "Le commissariat reçoit les plaintes des citoyens.",
        "La mairie accorde des permis de construire.",
        "Le cadastre enregistre les propriétés foncières.",
        "Les élections communales ont lieu tous les cinq ans.",
        "Le conseil municipal vote le budget annuel.",
        "La préfecture délivre les autorisations administratives.",
        "Le gouverneur coordonne les activités régionales.",
        "Les impôts fonciers financent les services publics.",
        "La douane contrôle les marchandises aux frontières.",
        "Le service de l'état civil reçoit le matin.",
        "La justice est rendue au nom du peuple.",
        "Le notaire authentifie les contrats officiels.",
        "Le registre du commerce immatricule les entreprises.",
        "La police municipale régule la circulation en ville.",
        "Le service des impôts calcule les taxes à payer.",
        "Les ambassades représentent les pays à l'étranger.",
        "La constitution garantit les droits fondamentaux.",
        "Le parlement adopte les lois du pays.",
        "Le président signe les décrets en conseil des ministres.",
        "La cour suprême est la plus haute juridiction.",
        "Les élus locaux gèrent les affaires de la commune.",
        "Le recensement permet de planifier le développement.",
        "L'état civil délivre les certificats de naissance.",
        "La chambre de commerce soutient les entreprises.",
        "Le ministère de l'éducation fixe les programmes scolaires.",
    ],
    "culture": [
        "Le griot chante les louanges du chef.",
        "Les femmes portent le bébé dans le dos.",
        "La cérémonie du mariage rassemble toute la famille.",
        "Les anciens transmettent la sagesse aux jeunes.",
        "Le tam-tam annonce une naissance dans le village.",
        "La danse du masque a lieu lors des funérailles.",
        "Les contes du soir éduquent les enfants.",
        "Le forgeron fabrique des outils en fer.",
        "La tisserande crée de beaux tissus colorés.",
        "Le potier façonne des jarres en argile.",
        "Les chants traditionnels accompagnent les travaux des champs.",
        "Le chef du village arbitre les conflits.",
        "Les rites de passage marquent la vie des individus.",
        "La musique traditionnelle est jouée aux cérémonies.",
        "Les jeunes apprennent les danses ancestrales.",
        "Le marché est un lieu de rencontre sociale.",
        "Les femmes se coiffent de manière traditionnelle.",
        "Les bijoux en or symbolisent la richesse.",
        "Le conteur fascine les enfants avec ses histoires.",
        "La langue maternelle est la première apprise.",
        "Les proverbes guident la conduite des hommes.",
        "Le sacrifice traditionnel honore les ancêtres.",
        "La natte tressée sert de lit dans les villages.",
        "Les cérémonies de circoncision rassemblent le village.",
        "Le chasseur connaît tous les secrets de la forêt.",
        "La pirogue traditionnelle navigue sur le fleuve.",
        "Les femmes brodent de magnifiques pagnes.",
        "Le berger mène son troupeau aux pâturages.",
        "Le bois sculpté représente les divinités ancestrales.",
        "La fête nationale célèbre l'indépendance du pays.",
    ],
    "economie": [
        "Le prix du carburant a augmenté cette semaine.",
        "Les femmes dominent le commerce informel local.",
        "La banque centrale régule la monnaie nationale.",
        "Les jeunes entrepreneurs créent des startups innovantes.",
        "L'agriculture représente la moitié du PIB national.",
        "Les minerais sont exportés vers les pays industriels.",
        "Le secteur des télécommunications se développe rapidement.",
        "Les routes bitumées facilitent le transport des marchandises.",
        "L'électricité est encore rare dans les zones rurales.",
        "Le taux de change influence les prix à l'importation.",
        "Les entreprises étrangères investissent dans les mines.",
        "Le marché financier est encore peu développé.",
        "Les femmes rurales s'organisent en tontines.",
        "La pauvreté touche surtout les zones rurales.",
        "L'aide internationale finance certains projets de développement.",
        "Le commerce équitable protège les petits producteurs.",
        "Les nouvelles technologies créent de nouveaux emplois.",
        "L'inflation érode le pouvoir d'achat des ménages.",
        "Les coopératives agricoles négocient de meilleurs prix.",
        "Le tourisme durable préserve l'environnement.",
        "Les transferts de la diaspora soutiennent l'économie.",
        "L'accès au crédit reste difficile pour les pauvres.",
        "La bonne gouvernance attire les investissements.",
        "Les marchés hebdomadaires animent les économies locales.",
        "Le secteur artisanal emploie des milliers de personnes.",
        "La dette publique pèse sur le budget de l'État.",
        "Les petites entreprises créent la majorité des emplois.",
        "L'assurance agricole protège les paysans des risques.",
        "La numérisation facilite les transactions financières.",
        "Le développement durable préserve les ressources naturelles.",
    ],
    "transport": [
        "Le bus relie la ville au village voisin.",
        "La moto-taxi est le transport le plus rapide en ville.",
        "Le chauffeur respecte les limitations de vitesse.",
        "La route est en mauvais état après les pluies.",
        "L'aéroport international accueille des vols quotidiens.",
        "Le taxi collectif prend cinq passagers maximum.",
        "Le pont traverse le fleuve en toute sécurité.",
        "Les camions transportent les marchandises vers le port.",
        "Le train de marchandises passe deux fois par semaine.",
        "La pirogue traverse la rivière en quelques minutes.",
        "Le carrefour est dangereux à cause de la circulation.",
        "Les feux de signalisation régulent la circulation.",
        "Le permis de conduire est nécessaire pour conduire.",
        "La voiture tombe en panne sur la route.",
        "Le mécanicien répare le moteur du camion.",
        "La gare routière est animée le matin.",
        "Le passager paie son billet avant de monter.",
        "L'ambulance transporte les malades urgents.",
        "La piste d'atterrissage est trop courte.",
        "Le navigateur guide le bateau sur le fleuve.",
    ],
    "environnement": [
        "La déforestation menace la biodiversité locale.",
        "Les déchets plastiques polluent les rivières.",
        "Le changement climatique affecte les saisons agricoles.",
        "La reforestation permet de lutter contre la désertification.",
        "L'eau potable manque dans de nombreux villages.",
        "Les feux de brousse détruisent la végétation.",
        "La sécheresse s'intensifie dans les régions sahéliennes.",
        "Les parcs naturels protègent la faune sauvage.",
        "Le recyclage des déchets préserve l'environnement.",
        "Les énergies renouvelables réduisent la pollution.",
        "Les inondations détruisent les récoltes et les maisons.",
        "La pollution de l'air cause des maladies respiratoires.",
        "Les puits permettent d'accéder à l'eau souterraine.",
        "La conservation des sols évite l'érosion.",
        "Les espèces menacées sont protégées par la loi.",
        "Le solaire fournit de l'électricité aux villages isolés.",
        "La gestion des eaux usées prévient les épidémies.",
        "Les arbres fruitiers nourrissent les communautés.",
        "La chasse abusive menace la faune sauvage.",
        "L'éducation environnementale sensibilise les jeunes.",
    ],
}

def enrichir_corpus_v2(tgt_code, tgt_name):
    # Chercher le fichier existant partout
    chemins_possibles = [
        FINAL_DIR / f"corpus_final_fr_{tgt_name}.json",
        FINAL_DIR / f"corpus_fr_{tgt_name}.json",
        Path("C:/Users/Albin Mabali/data/data/raw") / f"corpus_fr_{tgt_name}.json",
    ]

    existant = []
    for chemin in chemins_possibles:
        if chemin.exists():
            existant = json.loads(chemin.read_text(encoding="utf-8"))
            print(f"   Fichier trouvé : {chemin}")
            break

    existantes = {r["source_text"] for r in existant}
    print(f"{tgt_name.upper()} — Phrases existantes : {len(existant)}")

    nouvelles = []
    idx = len(existant)

    for domaine, phrases in phrases_supplementaires.items():
        for phrase in phrases:
            if phrase not in existantes:
                nouvelles.append({
                    "id": f"{tgt_code.upper()}_{idx:04d}",
                    "source_lang": "fr",
                    "target_lang": tgt_code,
                    "source_text": phrase,
                    "reference_translation": f"[À TRADUIRE EN {tgt_name.upper()}]",
                    "domain": domaine,
                    "source": "enrichissement_v3",
                    "grammatical_structure": "unknown",
                    "complexity": "simple",
                    "needs_translation": True,
                })
                existantes.add(phrase)
                idx += 1

    corpus_final = existant + nouvelles

    # Sauvegarder
    out = FINAL_DIR / f"corpus_final_fr_{tgt_name}.json"
    out.write_text(
        json.dumps(corpus_final, ensure_ascii=False, indent=2),
        encoding="utf-8"
    )

    print(f"   Nouvelles phrases ajoutées : {len(nouvelles)}")
    print(f"   Total : {len(corpus_final)}")
    print(f"   Sauvegardé → {out.name}")
    return corpus_final

# ── Lancement ────────────────────────────────────────
corpus_lin = enrichir_corpus_v2("lin", "lingala")
corpus_bam = enrichir_corpus_v2("bam", "bambara")

# ── Corpus combiné ────────────────────────────────────
tous = corpus_lin + corpus_bam
combined = FINAL_DIR / "corpus_combined_2000.json"
combined.write_text(
    json.dumps(tous, ensure_ascii=False, indent=2),
    encoding="utf-8"
)

print(f"\n{'='*50}")
print(f"  CORPUS FINAL")
print(f"  Lingala : {len(corpus_lin)} phrases")
print(f"  Bambara : {len(corpus_bam)} phrases")
print(f"  TOTAL   : {len(tous)} phrases")
print(f"{'='*50}")
```
