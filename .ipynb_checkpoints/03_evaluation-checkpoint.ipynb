```python
import os, json
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
from sacrebleu.metrics import BLEU, CHRF

# ── Chemins absolus corrigés ──────────────────────────
BASE_DIR    = Path(os.getcwd())
print(f"📁 Dossier actuel : {BASE_DIR}")

# Chercher translations_clean.json où qu'il soit
def find_file(filename, start=Path.home()):
    for p in start.rglob(filename):
        return p
    return None

trans_file = find_file("translations_clean.json")
print(f"📄 Fichier trouvé : {trans_file}")

RESULTS_DIR = trans_file.parent if trans_file else BASE_DIR
METRICS_DIR = BASE_DIR / "results" / "evaluations"
FIGURES_DIR = BASE_DIR / "results" / "figures"
METRICS_DIR.mkdir(parents=True, exist_ok=True)
FIGURES_DIR.mkdir(parents=True, exist_ok=True)

sns.set_theme(style="whitegrid")
plt.rcParams.update({"figure.dpi": 110, "font.size": 11})

df = pd.read_json(trans_file)
print(f"\n✅ {len(df)} traductions chargées")
print(df[["model","strategy","target_lang"]].value_counts())
```

```python
# ══════════════════════════════════════════════════════
#  CELLULE 2 — Calcul des métriques
# ══════════════════════════════════════════════════════
from sacrebleu.metrics import BLEU, CHRF

bleu_metric = BLEU(effective_order=True)
chrf_metric = CHRF()
results = []

for lang in ["lin", "bam"]:
    df_lang = df[df["target_lang"] == lang]

    for model in df_lang["model"].unique():
        for strategy in df_lang["strategy"].unique():

            subset = df_lang[
                (df_lang["model"]    == model) &
                (df_lang["strategy"] == strategy)
            ]
            if len(subset) < 5:
                continue

            hyps = subset["hypothesis"].tolist()
            refs = subset["reference"].tolist()

            # Filtrer placeholders
            pairs = [
                (h, r) for h, r in zip(hyps, refs)
                if r and "[À TRADUIRE]" not in str(r)
                and h and not str(h).startswith("ERREUR")
            ]
            if len(pairs) < 5:
                continue

            h_list = [p[0] for p in pairs]
            r_list = [[p[1] for p in pairs]]

            try:
                bleu = bleu_metric.corpus_score(h_list, r_list).score
                chrf = chrf_metric.corpus_score(h_list, r_list).score
            except:
                bleu, chrf = 0.0, 0.0

            results.append({
                "langue":    "Lingala" if lang=="lin" else "Bambara",
                "modele":    model,
                "strategie": strategy,
                "BLEU":      round(bleu, 2),
                "chrF":      round(chrf, 2),
                "n_phrases": len(pairs),
            })
            print(f"  {'LIN' if lang=='lin' else 'BAM'} | "
                  f"{model:12} | {strategy:10} | "
                  f"BLEU={bleu:5.2f} | chrF={chrf:5.2f} | n={len(pairs)}")

df_metrics = pd.DataFrame(results)

# Sauvegarde
out = METRICS_DIR / "metrics_auto.json"
out.write_text(
    json.dumps(results, ensure_ascii=False, indent=2),
    encoding="utf-8"
)
print(f"\n💾 Métriques sauvegardées → {out.name}")
print(f"\n{df_metrics.to_string(index=False)}")
```

```python
# ══════════════════════════════════════════════════════
#  CELLULE 3 corrigée — Visualisation chrF
# ══════════════════════════════════════════════════════
import matplotlib.pyplot as plt
import numpy as np

fig, axes = plt.subplots(1, 2, figsize=(14, 6))

for ax, langue in zip(axes, ["Lingala", "Bambara"]):
    subset = df_metrics[df_metrics["langue"] == langue].copy()
    subset["label"] = subset["modele"] + "\n" + subset["strategie"]
    subset = subset.sort_values("chrF", ascending=True)

    colors = ["#1B9E77" if "zero" in s else "#D95F02"
              for s in subset["strategie"]]

    bars = ax.barh(subset["label"], subset["chrF"],
                   color=colors, edgecolor="white", height=0.5)

    for bar, val in zip(bars, subset["chrF"]):
        ax.text(val + 0.01, bar.get_y() + bar.get_height()/2,
                f"{val:.2f}", va="center", fontweight="bold", fontsize=10)

    ax.set_xlim(0, 1.0)
    ax.set_xlabel("Score chrF (0 à 1)")
    ax.set_title(f"chrF — {langue}", fontsize=13)
    ax.axvline(0.5, color="red", linestyle="--",
               linewidth=1, label="Seuil 0.5")
    ax.legend(fontsize=8)

    from matplotlib.patches import Patch
    legend = [
        Patch(facecolor="#1B9E77", label="zero_shot"),
        Patch(facecolor="#D95F02", label="few_shot"),
    ]
    ax.legend(handles=legend, fontsize=9)

plt.suptitle(
    "Évaluation automatique chrF — LLaMA3-8b vs LLaMA3-70b\n"
    "Français → Lingala & Bambara",
    fontsize=12, fontweight="bold"
)
plt.tight_layout()
plt.savefig(str(FIGURES_DIR / "chrf_par_modele.png"), bbox_inches="tight")
plt.show()
print("✅ Graphique chrF sauvegardé !")
```
